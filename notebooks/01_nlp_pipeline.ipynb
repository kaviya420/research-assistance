```python
# ── PDF extraction ─────────────────────────────────────────
import pdfplumber        # Opens PDF files and extracts text page by page
                         # Better than PyPDF2 because it preserves layout

# ── Text processing ────────────────────────────────────────
import re                # Regular expressions — pattern matching in text
                         # Built into Python, no install needed
import string            # Contains punctuation constants like !"#$%&'()*+,-./:;<=>?

# ── NLP ────────────────────────────────────────────────────
import spacy             # Industrial NLP library
                         # Does tokenization, POS tagging, NER all at once
from keybert import KeyBERT   # Extracts keywords using BERT embeddings
                               # Smarter than simple word frequency counting

# ── Data handling ──────────────────────────────────────────
import json              # To pretty-print our output dictionary

print("All imports successful")
print(f"spaCy version: {spacy.__version__}")
```

```python
# Load spaCy's English model
# "en_core_web_sm" means:
#   en       = English language
#   core     = core capabilities (NER, POS, dependency parsing)
#   web      = trained on web text (news, blogs, Wikipedia)
#   sm       = small model (12MB) — fast, good enough for our use case
#   (alternatives: md=43MB, lg=741MB — more accurate but slower)

nlp = spacy.load("en_core_web_sm")

# Load KeyBERT model
# KeyBERT uses a sentence transformer model internally
# It converts words and sentences into vectors
# then finds which words are most similar to the whole document's meaning
# those words = keywords
kw_model = KeyBERT()

print("spaCy model loaded:", nlp.meta['name'])
print("KeyBERT model loaded successfully")
print()

# Let's see what spaCy can detect in a simple test sentence
test_sentence = "BERT model achieves state-of-the-art results on the SQuAD dataset by Google."
doc = nlp(test_sentence)

print("Named Entities in test sentence:")
for ent in doc.ents:
    print(f"  '{ent.text}' → Type: {ent.label_} ({spacy.explain(ent.label_)})")
```
